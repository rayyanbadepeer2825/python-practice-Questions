## Section D: Loops
**Q10. Print the multiplication table (1 to 10) of any given number using a for loop.**

**Q11. Take n as input and find the sum of numbers from 1 to n using a while loop.**

```python
n = int(input("Enter n: "))
total_sum = 0
i = 1

while i <= n:
    total_sum += i
    i += 1
print(f"Sum={total_sum}")
```

**Q13. Take a number n as input and find the sum of its digits using a while loop.**

```python
n = int(input("Enter n: "))
sum_digits = 0

while n > 0:
    sum_digits += n % 10
    n //= 10
print(f"Sum of digits = {sum_digits}")
```

**Q14. Take n as input and print the first n terms of the Fibonacci series.**

```python
n = int(input("Enter n: "))
a, b = 0, 1

for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

## Section E: Classic Combined Problems
**Q15. Palindrome Check**

```python
n = int(input("Enter n: "))
original = n
rev = 0

while n > 0:
    rev = (rev * 10) + (n % 10)
    n //= 10
    
if original == rev:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

**Q16. Armstrong Number**

```python
n = int(input("Enter n: "))
original = n
sum_arm = 0

while n > 0:
    digit = n % 10
    sum_arm += digit ** 3
    n //= 10
    
if sum_arm == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

**Q17. Reverse a Number (without string conversion)**

```python
n = int(input("Enter n: "))
rev = 0

while n > 0:
    rev = (rev * 10) + (n % 10)
    n //= 10
print(f"Reversed number = {rev}")
```

**Q18. Prime Number Check**

```python
n = int(input("Enter n: "))

if n <= 1:
    print("Not a Prime Number")
else:
    is_prime = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            is_prime = False
            break
            
    if is_prime:
        print("Prime Number")
    else:
        print("Not a Prime Number")
```

**Q19. GCD of Two Numbers using a loop**

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))

while b != 0:
    a, b = b, a % b
print(f"GCD={a}")
```
